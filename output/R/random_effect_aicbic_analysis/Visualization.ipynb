### This notebook is used to visualize the data from random_effect_analysis

```python
import os

curr_dir = os.getcwd()
AIC_path = os.path.join(curr_dir, "graph", "AIC")
if not os.path.exists(AIC_path):
    os.makedirs(AIC_path)
BIC_path = os.path.join(curr_dir, "graph", "BIC")
if not os.path.exists(BIC_path):
    os.makedirs(BIC_path)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read your CSV (adjust the path if needed)
df = pd.read_csv('random_effect_aicbic_result.csv')

# Get all experiments
experiments = df['experiment'].unique()

for exp in experiments:
    # Filter data for this experiment
    sub = df[df['experiment'] == exp]

    # Pivot so that:
    # - index: dimensions (x-axis)
    # - columns: random_effect (different bars per dimension)
    # - values: AIC scores (y-axis)
    pivot = sub.pivot(index='dimension', columns='random_effect', values='AIC')

    # Create bar chart
    fig, ax = plt.subplots()
    pivot.plot(kind='bar', ax=ax)

    # Labels and title
    ax.set_title(exp)
    ax.set_xlabel("Dimension")
    ax.set_ylabel("AIC Score")

    # Make layout nicer
    fig.tight_layout()

    # Save figure (optional)
    out_path = os.path.join(AIC_path, f"{exp}.png")
    fig.savefig(out_path)

    # Close to free memory when looping
    plt.close(fig)

for exp in experiments:
    # Filter data for this experiment
    sub = df[df['experiment'] == exp]

    # Pivot so that:
    # - index: dimensions (x-axis)
    # - columns: random_effect (different bars per dimension)
    # - values: BIC scores (y-axis)
    pivot = sub.pivot(index='dimension', columns='random_effect', values='BIC')

    # Create bar chart
    fig, ax = plt.subplots()
    pivot.plot(kind='bar', ax=ax)

    # Labels and title
    ax.set_title(exp)
    ax.set_xlabel("Dimension")
    ax.set_ylabel("BIC Score")

    # Make layout nicer
    fig.tight_layout()

    # Save figure (optional)
    out_path = os.path.join(BIC_path, f"{exp}.png")
    fig.savefig(out_path)

    # Close to free memory when looping
    plt.close(fig)
```
