### This notebook is used to visualize the data from random_effect_analysis

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

def process_algo(data, algo):
    path = os.path.join(data, algo, "aic_bic_result.csv")

    curr_dir = os.getcwd()
    AIC_path = os.path.join(curr_dir, data, algo, "graph", "AIC")
    if not os.path.exists(AIC_path):
        os.makedirs(AIC_path)
    BIC_path = os.path.join(curr_dir, data, algo, "graph", "BIC")
    if not os.path.exists(BIC_path):
        os.makedirs(BIC_path)

    # Read your CSV (adjust the path if needed)
    df = pd.read_csv(path)

    # Get all experiments
    experiments = df['experiment'].unique()

    for exp in experiments:
        # Filter data for this experiment
        sub = df[df['experiment'] == exp]

        # Pivot so that:
        # - index: dimensions (x-axis)
        # - columns: random_effect (different bars per dimension)
        # - values: AIC scores (y-axis)
        pivot = sub.pivot(index='dimension', columns='random_effect', values='AIC')
        
        # Calculate difference from baseline (none)
        if 'none' in pivot.columns:
            baseline = pivot['none']
            diff_pivot = pivot.drop('none', axis=1).sub(baseline, axis=0)
        else:
            diff_pivot = pivot

        # Create bar chart
        fig, ax = plt.subplots()
        diff_pivot.plot(kind='bar', ax=ax)

        # Labels and title
        ax.set_title(f"{exp} (AIC difference from baseline 'none')")
        ax.set_xlabel("Dimension")
        ax.set_ylabel("AIC Difference from Baseline")
        ax.axhline(y=0, color='black', linestyle='--', linewidth=0.8)

        # Make layout nicer
        fig.tight_layout()

        # Save figure (optional)
        out_path = os.path.join(AIC_path, f"{exp}.png")
        fig.savefig(out_path)

        # Close to free memory when looping
        plt.close(fig)

    for exp in experiments:
        # Filter data for this experiment
        sub = df[df['experiment'] == exp]

        # Pivot so that:
        # - index: dimensions (x-axis)
        # - columns: random_effect (different bars per dimension)
        # - values: BIC scores (y-axis)
        pivot = sub.pivot(index='dimension', columns='random_effect', values='BIC')
        
        # Calculate difference from baseline (none)
        if 'none' in pivot.columns:
            baseline = pivot['none']
            diff_pivot = pivot.drop('none', axis=1).sub(baseline, axis=0)
        else:
            diff_pivot = pivot

        # Create bar chart
        fig, ax = plt.subplots()
        diff_pivot.plot(kind='bar', ax=ax)

        # Labels and title
        ax.set_title(f"{exp} (BIC difference from baseline 'none')")
        ax.set_xlabel("Dimension")
        ax.set_ylabel("BIC Difference from Baseline")
        ax.axhline(y=0, color='black', linestyle='--', linewidth=0.8)

        # Make layout nicer
        fig.tight_layout()

        # Save figure (optional)
        out_path = os.path.join(BIC_path, f"{exp}.png")
        fig.savefig(out_path)

        # Close to free memory when looping
        plt.close(fig)
```

```python
algos = ["NLD", "ScaSim", "MultiMatch"]
data_set = ["EMIP_corrected", "Code_rendering"]

for data in data_set:
    for algo in algos:
        process_algo(data, algo)
```
