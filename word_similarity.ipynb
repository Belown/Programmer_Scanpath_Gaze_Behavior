```python
from gensim.models import KeyedVectors
from scipy.stats import spearmanr

# Load pre-trained Google Word2Vec
model = KeyedVectors.load_word2vec_format(
    'GoogleNews-vectors-negative300.bin.gz',
    binary=True
)

# Load your human dataset, e.g. WordSim-353
pairs = []
human_scores = []

with open('wordsim353.tsv', 'r', encoding='utf-8') as f:
    next(f)  # skip header
    for line in f:
        w1, w2, score = line.strip().split('\t')
        if w1 in model.key_to_index and w2 in model.key_to_index:
            pairs.append((w1, w2))
            human_scores.append(float(score))

# Compute cosine similarities from model
model_scores = [model.similarity(w1, w2) for w1, w2 in pairs]

# Evaluate correlation with human ratings
corr, _ = spearmanr(model_scores, human_scores)
print(f"Spearman correlation: {corr:.3f}")
```
