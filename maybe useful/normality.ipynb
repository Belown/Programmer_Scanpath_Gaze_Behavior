```python
import pandas as pd
from scipy.stats import shapiro
import matplotlib.pyplot as plt
import seaborn as sns

# Load your data (replace with your filename)
df = pd.read_csv(r"D:\Storage\ETH\Thesis\Code\notebook\EMIP\comparison_results\between_group\trial_5\Java_high.csv_vs_Java_none.csv_results.csv")

# Define the five dimensions
dimensions = ["Shape", "Length", "Direction", "Position", "Duration"]

# Run Shapiro-Wilk test for each
print("Shapiro-Wilk normality test results:\n")
for dim in dimensions:
    stat, p = shapiro(df[dim])
    print(f"{dim:10s}  W = {stat:.4f},  p = {p:.4f}")
    if p > 0.05:
        print(" → Likely normal ✅\n")
    else:
        print(" → Not normal ❌\n")

# Optional: visualize distributions
for dim in dimensions:
    plt.figure(figsize=(6,4))
    sns.histplot(df[dim], kde=True)
    plt.title(f"{dim} distribution (within-group)")
    plt.show()
```
