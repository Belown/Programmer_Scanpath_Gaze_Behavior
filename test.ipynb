```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 读取数据
data = pd.read_csv("comparison_results/high_results.csv")

# 选择一个分数类型（例如 Shape）
pivot_data = data.pivot("exp_a", "exp_b", "Shape")

# 绘制热力图
plt.figure(figsize=(10, 8))
sns.heatmap(pivot_data, annot=True, cmap="coolwarm", fmt=".4f")
plt.title("Shape Score Heatmap")
plt.show()
```
